Hipótese nula (H₀): A duração média das corridas não muda entre sábados chuvosos e sábados com bom tempo.

Hipótese alternativa (H₁): A duração média das corridas muda entre sábados chuvosos e sábados com bom tempo.

Vou usar o teste t de Student para amostras independentes (scipy.stats.ttest_ind) porque:

Estamos comparando médias entre dois grupos.
As amostras são independentes (cada corrida é única).
A variável de interesse (duration_seconds) é quantitativa contínua.

<div class="alert alert-block alert-danger">
<b> Comentário do revisor: </b> <a class="tocSkip"></a>
<s>
    
Aqui você precisa calcular a variância para dias chuvosos e dias não chuvoses para definir o argumento `equal_var`. Você precisa usar o resultado dessa análise para definir que as variâncias são diferentes (argumento `equal_var=False`)
    
```python
# descobrir se as variâncias são diferentes
good.var(), bad.var()

```    
<br>
    
**Sugestão:** Você pode fazer o teste no código ou apenas imprimir as variâncias e concluir se são iguais ou diferentes. Essa etapa é importante para satisfazer a premissa do teste-t
</s>
</div>

````python
import pandas as pd
from scipy.stats import ttest_ind

#Importar os dados
df = pd.read_csv('/datasets/project_sql_result_07.csv', parse_dates=['start_ts'])

#Separar os grupos
bad = df[df['weather_conditions'] == 'Bad']['duration_seconds']
good = df[df['weather_conditions'] == 'Good']['duration_seconds']

#Calcular variâncias
var_bad = bad.var()
var_good = good.var()

print("Variância - Bad:", var_bad)
print("Variância - Good:", var_good)

#Teste t de Student (Welch’s t-test, pois variâncias são diferentes)
t_stat, p_value = ttest_ind(bad, good, equal_var=False)

#Nível de significância
alpha = 0.05

#Conclusão
print(f"t = {t_stat:.2f}, p = {p_value:.15f}")
if p_value < alpha:
    print("Rejeitamos H₀: há diferença significativa na duração média.")
else:
    print("Não rejeitamos H₀: não há diferença significativa na duração média.")
````

<div class="alert alert-block alert-danger">
<b> Comentário do revisor: </b> <a class="tocSkip"></a>

</div>

# Conclusão
Como as variâncias são diferentes, usamos Welch’s t-test (equal_var=False).
Como o valor-p é menor que o nível de significância (p < 0.05), rejeitamos a hipótese nula. Isso indica que há uma diferença estatisticamente significativa na duração média das corridas entre sábados com clima bom e sábados chuvosos.
